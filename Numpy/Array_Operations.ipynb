```python
mat1 = np.array([[1, 2], [3, 4]])
mat2 = np.array([[5, 6], [7, 8]])
```

```python
# stacking array
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
```

```python
np.vstack((arr1, arr2))
```

```python
np.hstack((arr1, arr2))
```

```python
np.column_stack((arr1, arr2))
```
